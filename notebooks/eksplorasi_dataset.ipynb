# Eksplorasi Dataset Social Media Engagement

Notebook ini digunakan di JupyterLab untuk memahami data, memeriksa kualitas data, dan menghasilkan ringkasan awal sebelum menjalankan dashboard Streamlit.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd()
if not (ROOT / "data").exists() and (ROOT.parent / "data").exists():
    ROOT = ROOT.parent
DATA_PATH = ROOT / "data" / "social_media_engagement_clean.csv"
df = pd.read_csv(DATA_PATH, parse_dates=["timestamp"])
df.head()
```

## 1. Ukuran dan tipe data

```python
print("Ukuran dataset:", df.shape)
display(df.dtypes.to_frame("Tipe data"))
display(df.describe(include="all").T)
```

## 2. Pemeriksaan kualitas data

```python
display(df.isna().sum().sort_values(ascending=False).to_frame("Jumlah missing"))
print("Jumlah baris duplikat:", df.duplicated().sum())
print("Rentang waktu:", df["timestamp"].min(), "sampai", df["timestamp"].max())
```

## 3. Ringkasan kategori

```python
for col in ["platform", "sentiment_label", "topic_category"]:
    print(f"\\nDistribusi {col}")
    display(df[col].value_counts().head(15).to_frame("Jumlah"))
```

## 4. KPI dan agregasi awal

```python
df["total_interactions"] = df[["likes_count", "shares_count", "comments_count"]].fillna(0).sum(axis=1)
print("Total interaksi:", f'{df["total_interactions"].sum():,.0f}')
print("Rata-rata engagement rate:", f'{df["engagement_rate"].mean():.2%}')
print("Total impressions:", f'{df["impressions"].sum():,.0f}')
display(df.groupby("platform")["engagement_rate"].mean().sort_values(ascending=False).to_frame("Rata-rata engagement rate"))
```

## 5. Grafik eksplorasi

```python
df["sentiment_label"].value_counts().plot(kind="bar", title="Distribusi Sentimen", ylabel="Jumlah postingan")
plt.tight_layout()
plt.show()

df.groupby("platform")["engagement_rate"].mean().sort_values().plot(kind="barh", title="Rata-rata Engagement Rate per Platform", xlabel="Engagement rate")
plt.tight_layout()
plt.show()
```
